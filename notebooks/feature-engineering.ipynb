```python
import sys
import os
sys.path.append(os.path.abspath('../'))

from src.data_preprocessing import load_data, clean_data, merge_with_ip
from src.feature_engineering import add_features, prepare_data
```

```python
fraud_raw, ip_raw = load_data('../data/raw/Fraud_Data.csv', '../data/raw/IpAddress_to_Country.csv')
df = clean_data(fraud_raw)
df = merge_with_ip(df, ip_raw)
df = add_features(df)
```

```python
X_train, X_test, y_train, y_test = prepare_data(df)

# Save to processed folder
os.makedirs('../data/processed/', exist_ok=True)
X_train.to_csv('../data/processed/X_train_final.csv', index=False)
y_train.to_csv('../data/processed/y_train_final.csv', index=False)
X_test.to_csv('../data/processed/X_test_final.csv', index=False)
y_test.to_csv('../data/processed/y_test_final.csv', index=False)
```
